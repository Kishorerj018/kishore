```python
class Transaction:
    def __init__(self, amount, category, note, trans_type, date=None):
        self.__amount = amount               # Private should not be accessed directly outside
        self._category = category            # Protected accessible for class and subclasses
        self.note = note                     # Public
        self.__trans_type = trans_type       # Private
        self.date = date                     

    def str(self):  
        return (f"{self.date} | {self.__trans_type} | {self._category} | {self.__amount} | {self.note}")

    def get_amount(self):
        return self.__amount

    def get_type(self):
        return self.__trans_type


class Account:
    def __init__(self, owner):
        self.owner = owner               # Public
        self._transactions = []          # Protected list of transactions

    def add_transaction(self, transaction):
        try:
            transaction.get_amount()
            transaction.get_type()
            self._transactions.append(transaction)
        except AttributeError:
            print("enter a valid Transaction")

    def get_balance(self):
        balance = 0
        for t in self._transactions:
            if t.get_type() == 'Income':
                balance += t.get_amount()
            else:
                balance -= t.get_amount()
        return balance

    def transactions_by_category(self, category):
        results = ()
        for t in self._transactions:
            if t._category == category:
                results.append(t)
        return results


class Report:
    def __init__(self, account):
        self._account = account   

    def category_summary(self):
        summ = {}
        for t in self._account._transactions:
            if t._category in summ:
                pass
            else:
                summ[t._category] = 0

            if t.get_type() == 'Income':
                summ[t._category] += t.get_amount()
            else:
                summ[t._category] -= t.get_amount()
        return summ

    def spend_money(self):
        total= 0
        for t in self._account._transactions:
            if t.get_type() == "Expense":
                total += t.get_amount()
        return total


kumar = Account("kishore kumar")
kumar.add_transaction(Transaction(50000, "Salary", "August Salary", "Income", "2025-09-05"))
kumar.add_transaction(Transaction(7000, "Food", "Groceries", "Expense", "2025-09-06"))
kumar.add_transaction(Transaction(2000, "Transport", "Bus passes", "Expense", "2025-09-08"))
kumar.add_transaction(Transaction(5000, "house bills", "current", "Expense", "2025-09-11"))
kumar.add_transaction(Transaction(7000, "hospital", "medicines", "Expense", "2025-09-15"))
kumar.add_transaction(Transaction(2000, "mobile bill", "recharges", "Expense", "2025-09-16"))

print("Category amount:")
santhosh = Report(a1)  
rowdy= santhosh.category_summary()
print(rowdy)


print("\nAll Transactions:")
for t in kumar._transactions:
    print(t.str())

balance = kumar.get_balance()
print("Account Balance is:", balance)
spent = kishore.spend_money()
print("Total Money Spent is:", spent)
```
