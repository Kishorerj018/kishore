```python
#single level inheritance
class sla():   #parent class or base class
    def dataanalytics(self):
        print("powerbi,python,sql")
    def pyt(self):
        print("html,css,python")
    def cloud(self):
        print("aws and azure")

class navalur(sla):   #child class or derived class
    def data_science(self):
        print("al,ml and ai")

kishore=navalur()
kishore.cloud()
```

```python
#hierachial inheritance
class sla():   #parent class or base class
    def dataanalytics(self):
        print("powerbi,python,sql")
    def pyt(self):
        print("html,css,python")
    def cloud(self):
        print("aws and azure")

class navalur(sla):   #child class or derived class
    def data_science(self):
        print("al,ml and ai")

class placements():
    def fre(self):
        print("join as fresher")
    def exp(self):
        print("join as exp")

class palavaram(sla):  #child class
    def ccn(self):
        print("networking")

class kknagar(sla,placements):  #multiple inheritance
    def comm(self):
         print("learn comm")

class anna_nagar(navalur): #multi - level inheritance
    def apptitute(self):
          print("learn appti")

kishore=kknagar()
kishore.fre()
```

```python
#calculator -----a b c d
#add ---a b
#sub ---c>d c-d d>c ---d-c
#mult ----c * d
class calculator():
    def __init__(self,a,b,c,d):
        self.a=a
        self.b=b
        self.c=c
        self.d=d
    def add(self):
        print(self.a + self.b)
    def sub(self):
        if self.c>self.d:
            print(self.c-self.d)
        else:
            print(self.d-self.c)


class calc2(calculator):
    def __init__(self,a,b,c,d):
        super().__init__(a,b,c,d)
    def multi(self):
        super().add()
        print(self.c*self.d)

kishore=calc2(10,20,30,40)
kishore.multi()
```

```python
class Employee:
    def __init__(self, name, age, gender, location="", base=10000, salary=0):
        self.name = name
        self.age = age
        self.gender = gender
        self.location = location
        self.base = base
        self.salary = salary

    def company(self):
        print("Our company name is Lenovo")

    def output(self):
        print(f"My name is {self.name}my age is {self.age} and my gender is {self.gender}")

class FullTimeSalary(Employee):
    def __init__(self, name, age, gender, location="tiruttani", base=10000, bonus=0):
        super().__init__(name, age, gender, location, base)
        self.bonus = bonus

    def salary_calc(self):
        print (self.base + self.bonus)

    

class PartTimeSalary(Employee):
    def __init__(self, name, age, gender, location="tiruttani", hours=0, hour_pay=100):
        super().__init__(name, age, gender, location, base=0) 
        self.hours = hours
        self.hour_pay = hour_pay

    def salary_cal(self):
        print (self.hours * self.hour_pay)
--
    


dilli = FullTimeSalary("Dilli", 25, "Male", bonus=5000)
dilli.output()
dilli.salary_calc()

arun = PartTimeSalary("arun", 22, "Male", hours=700)
dilli.output()
arun.salary_cal()
    
```

```python
class Cars:
    def __init__(self, model, price=2000, service=0, year=0):
        self.model = model
        self.price = price
        self.service = service
        self.year = year

    def output(self):
        print(f"Car model is {self.model} is Price {self.price} is Service {self.service}")


class Cars2(Cars):
    def __init__(self, model, price, charge=3000, service=0, year=0):
        super().__init__(model, price, service, year) 
        self.charge = charge

    def service_co(self):
        print( self.charge + self.price)


class Cars3(Cars):
    def __init__(self, model, price=2000, service=0, year=0, hours=1, hours_pay=2000):
        super().__init__(model, price, service, year) 
        self.hours = hours
        self.hours_pay = hours_pay

    def service_coo(self):
        print(self.hours * self.hours_pay)


tata = Cars2("Rolls Royce", 2000, 3000)
tata.output()
tata.service_co()

bmw = Cars3("Benz", 5, 1500)
bmw.output()
bmw.service_coo()
```
