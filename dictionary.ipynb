```python
name="JJJjjJBBBbb"
alpha={}
for i in name:
    if i.isupper():
        if i in alpha:
            alpha[i]=alpha[i]+1
        else:
            alpha[i]=1
print(alpha)
```

```python
str1="AAA aaa aB"
alpha={}
for i in str1:
    if i==" ":
        continue
    if i in alpha:
        alpha[i]=alpha[i]+1
    else:
        alpha[i]=1
print(alpha)
```
