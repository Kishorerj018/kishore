```python
class Counselor:
    def __init__(self, name, age, phone, degree, exp, fees, status):
        self.name = name
        self.age = age
        self.degree = degree
        self.exp = exp
        self.fees = fees
        self.status = status

        try:
            if len(phone) == 10 and (phone.startswith(("6", "7", "8", "9"))):
                self.phone = phone
                self.valid = True
            else:
                raise ValueError
        except ValueError:
            print("Invalid phone number")
            self.valid = False

    def course(self):
        if self.valid:
            course = input("Enter course: ")
            self.selected_course = course
            if self.selected_course.lower() == "data analyst":
                print("PowerBI, Python, and SQL")
            elif self.selected_course.lower() == "data science":
                print(" ML, AI, and DL")
            elif self.selected_course.lower() == "cloud":
                print(" AWS and Azure")
            elif self.selected_course.lower() == "pyfs":
                print("HTML, CSS, JS, SQL, Bootstrap, Python, Django")
            else:
                print("Enter valid course")

    def training(self):
        print("We provide 6 four-hour sessions of training.")


class Trainer(Counselor):
    def __init__(self, name, age, phone, degree, status):
        super().__init__(name, age, phone, degree, exp="NA", fees=0, status=status)
        trainename = input("Enter trainer name: ")
        self.trainename = trainename.lower()

        if self.trainename == "santhosh":
            print(f"Hello {self.name}, your trainer is {self.trainename}.")
            print("Python Full-Stack, Frontend")
        elif self.trainename == "selva":
            print(f"Hello {self.name}, your trainer is {self.trainename}.")
            print("Machine Learning, Deep Learning, AI")
        elif self.trainename == "kishore":
            print(f"Hello {self.name}, your trainer is {self.trainename}.")
            print("Django and Flask")
        elif self.trainename == "kumar":
            print(f"Hello {self.name}, your trainer is {self.trainename}.")
            print("Java Full-Stack and MERN")
        else:
            print("Enter valid trainer")


class Placements(Counselor):
    def __init__(self, name, age, degree, status):
        self.name = name
        self.age = age
        self.degree = degree
        self.status = status

    def fresher(self):
        print("We place freshers.")

    def experience(self):
        print("We place experienced professionals.")



name = input('Enter your name: ')
age = int(input('Enter your age: '))
phone = input('Enter your phone number: ')
degree = input('Enter your degree: ')
exp = input('Enter your experience: ')
fees = int(input('Enter your fees: '))
status = input('Enter your status: ')


c1 = Counselor(name, age, phone, degree, exp, fees, status)
c1.course()
c1.training()


t1 = Trainer(name, age, phone, degree, status)

p1 = Placements(name, age, degree, status)
p1.fresher()
p1.experience()
```

```python
from datetime import datetime

# ----------------------
# Task Class
# ----------------------
class Task:
    def __init__(self, title, description, due_date=None):
        # Private attributes
        self.__title = title
        self.__completed = False
        
        # Protected attribute
        self._description = description
        
        # Public attribute
        self.due_date = due_date if due_date else datetime.now()

    # String representation
    def __str__(self):
        status = "✔️ Done" if self.__completed else "❌ Pending"
        return f"{self.__title} | {self._description} | Due: {self.due_date.date()} | {status}"

    # ---------- Getter & Setter ----------
    def get_title(self):
        return self.__title

    def set_title(self, new_title):
        try:
            if not new_title.strip():
                raise ValueError("Title cannot be empty!")
            self.__title = new_title
        except ValueError as e:
            print(f"Error: {e}")

    def is_completed(self):
        return self.__completed

    def mark_completed(self):
        self.__completed = True

    def mark_pending(self):
        self.__completed = False


# ----------------------
# ToDoList Class
# ----------------------
class ToDoList:
    def __init__(self, owner):
        # Public
        self.owner = owner
        # Protected
        self._tasks = []

    def add_task(self, task):
        try:
            if isinstance(task, Task):
                self._tasks.append(task)
            else:
                raise TypeError("Only Task objects can be added!")
        except TypeError as e:
            print(f"Error: {e}")

    def remove_task(self, title):
        try:
            for t in self._tasks:
                if t.get_title() == title:
                    self._tasks.remove(t)
                    print(f"Task '{title}' removed.")
                    return
            raise ValueError("Task not found!")
        except ValueError as e:
            print(f"Error: {e}")

    def get_pending_tasks(self):
        return [t for t in self._tasks if not t.is_completed()]

    def get_completed_tasks(self):
        return [t for t in self._tasks if t.is_completed()]


# ----------------------
# Report Class
# ----------------------
class Report:
    def __init__(self, todo_list):
        self._todo_list = todo_list

    def summary(self):
        total = len(self._todo_list._tasks)
        completed = len(self._todo_list.get_completed_tasks())
        pending = len(self._todo_list.get_pending_tasks())
        
        return {
            "owner": self._todo_list.owner,
            "total_tasks": total,
            "completed_tasks": completed,
            "pending_tasks": pending
        }


# ----------------------
# Example Usage
# ----------------------
if __name__ == "__main__":
    my_list = ToDoList("Alice")

    # Add tasks
    my_list.add_task(Task("Buy groceries", "Milk, Bread, Eggs", datetime(2025, 9, 20)))
    my_list.add_task(Task("Finish project", "Submit assignment", datetime(2025, 9, 25)))
    my_list.add_task(Task("Call Bob", "Discuss weekend plan"))

    # Mark one task completed
    my_list._tasks[0].mark_completed()

    # Remove task
    my_list.remove_task("Call Bob")

    # Create report
    report = Report(my_list)
    print("\nSummary:", report.summary())

    # Show all tasks
    print("\nAll Tasks:")
    for t in my_list._tasks:
        print(t)

    # Show only pending tasks
    print("\nPending Tasks:")
    for t in my_list.get_pending_tasks():
        print(t)
```
